# CS549 Project — Exploratory Data Analysis

Quick look at the merged financial-transaction dataset produced by `scripts/build_dataset.py`. The notebook is intentionally lightweight; treat it as a starting point for the figures that go into the final report.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import pandas as pd

from src import config

df = pd.read_csv(config.MERGED_CLEAN_CSV, parse_dates=["date"])
df.head()
```

## Class distribution

```python
ax = (
    df["category"]
    .value_counts()
    .sort_values()
    .plot.barh(figsize=(8, 6), color="#3c79b8")
)
ax.set_xlabel("Number of transactions")
ax.set_title("Class distribution after taxonomy unification")
plt.tight_layout()
plt.show()
```

## Source breakdown

```python
(
    df.groupby(["category", "source"]).size().unstack(fill_value=0)
    .plot.bar(stacked=True, figsize=(10, 5), colormap="tab10")
)
plt.ylabel("Number of transactions")
plt.title("Per-class contribution by source dataset")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

## Amount distribution per category

```python
fig, ax = plt.subplots(figsize=(10, 5))
df.boxplot(column="amount", by="category", ax=ax, rot=30, grid=False, showfliers=False)
ax.set_title("Transaction amount by category (outliers hidden)")
ax.set_xlabel("")
ax.set_ylabel("Amount (USD)")
plt.suptitle("")
plt.tight_layout()
plt.show()
```

## Top descriptions per category

```python
for cat, group in df.groupby("category"):
    top = group["description"].value_counts().head(5)
    print(f"\n{cat}")
    print(top.to_string())
```
